##### Q9. Find the Missing Number from 1 to N
Write a Python program to find the missing number in a list containing numbers from 1 to N.
The list contains exactly one missing number, and all remaining numbers appear only once.
Example Input:
nums = [1, 2, 4, 5, 6]
N = 6
Expected Output:
3

##### pseduo code
- all these include in function missing_number()
1. take n  variable to calculate the len(nums)+1
2. calculate the expected_sum by using formula n*(n+1)//2
3. then calculate the actual_sum of the list nums
4. to find the missing_value difference between expected_sum and actual_sum
5. return the missing_value  

```python
def missing_number():
    n=len(nums)+1
    expected_sum=n*(n+1)//2
    actual_sum=sum(nums)
    missing_value=expected_sum - actual_sum
    return missing_value
nums=[1,2,4,5,6]
print(missing_number())
```

##### Q10. Find Two Numbers Whose Sum Equals a Target
Write a Python program to find the first pair of numbers whose sum equals a given target.
Return the pair as a list. If no valid pair exists, return an empty list.
Return the pair whose second element is encountered earliest when scanning the input from left to right. If multiple earlier elements qualify, choose the earliest one.
Example Input:
nums = [2, 7, 11, 15]
target = 9
Expected Output:
[2, 7]

##### pseduo code
- all these steps in a function target_sum()
1. take a complement variable ,empty set and empty list
2. iterate throught the list given 
3. calculate the complement by subtracting the number from target 
4. add that to the empty set 
5. when ever the complement in the set then add that complement and iterate value in list by using extend adds multiple values to the list  
6. return the list 

```python
def target_sum(l,target):
    s=set()
    pair=[]
    for i in l:
        complement=target-i
        if complement in s:
            pair.extend((complement,i))
        s.add(i)
    return pair
l=[2,7,11,15]
target=9
print(target_sum(l,target))
```

##### Q12. Find Common Elements Between Two Lists
Write a Python program to find the common elements between two lists.
Each common element should appear only once in the result, maintaining the order of its first occurrence in the first list.
Example Input:
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
Expected Output:
[3, 4, 5]

##### pseduo code
- all these steps include in function common_values()
1. take an empty list to store the common values 
2. iterate over the elements in both the lists by nested loops
3. compare the values are equal if  condition is true and that are not in list 
4. then append the values in list 
5. return the list 

```python
def common_values(l1,l2):
    common=[]
    for i in l1:
        for j in l2:
            if i==j:
                if i not in common:
                    common.append(i)
    return common
l1=[1,2,3,4,5]
l2=[3,4,5,6,7]
print(common_values(l1,l2))
```

Assume the list contains at least two elements.
Example Input:
[4, 7, 2, 9, 5]
Expected Output:
14

##### pseduo code
- all these steps include in function consective_sum()
1. take max_sum as sum of first two elemnets in the list 
2. take current_sum variable as 
3. iterate throught the list calculate the current_sum by the iterate value and the next value 
4. check if the current_sum greater than the max_sum 
5. then update the max_sum 
6. return the max_sum

```python
def consecutive_sum(lis):
    max_sum=lis[0]+lis[1]
    curren_sum=0
    for i in range(1,len(lis)-1):
        curren_sum=lis[i]+lis[i+1]
        if curren_sum>max_sum:
            max_sum=curren_sum
    return max_sum
l=[4,7,2,9,5]
print(consecutive_sum(l))
```

Return each pair in ascending numerical order. Return the list of pairs sorted by their first element. If no valid pairs exist, return an empty list.

4. add that to the empty set
