## Q35 output --> [[1,99],[3,4]]

```python
import copy
a = [[1, 2], [3, 4]]     # create list a
b = copy.copy(a)         # create a new outer list and nested list are shared b/w a & b
b[0][1] = 99             # modifies the nested list
print(a)                 # changes occur in both so a is [[1,99],[3,4]]
```

```python
print(a)
print(b)
print(id(a))
print(id(b))
```
