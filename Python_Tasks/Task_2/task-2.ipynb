### Q1. Find the Shortest Word in a Sentence
Write a Python program to find the shortest word in a given sentence.

### Q8. Find the Most Frequent Character
Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.
Example input:
banana
Expected output:
a

- observe = repeating character in the given string
- understand = first most repeating character 
- break =  dictionary 
- Logic =  for loop and dictionary and break after getting first most character
- code =
- test = 'Innomatics' i

```python
s='banana'.lower()
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
most_char=max(d.values())
for key,value in d.items():
    if value==most_char:
        print('Most Frequent Character:',key)
        break
```

### Q20. Find the Longest Word That Appears Only Once
Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.
Example input:
Python makes coding useful Python
Expected output:
coding


- observe = find longest word and excatly appear once 
- understand =  longest word that appear only once in a sentence
- break =  loop,dict
- Logic = for loop
- code =
- test = 'I am Learning python programming and interseting'

```python
s='Python makes coding useful Python'.split()
d={}
for i in s:
    if i in d:
        d[i]+=1
    else:
        d[i]=1
longest=''
for key,value in d.items():
    if value==1 and len(key)>len(longest):
        longest=key
print(longest)
```
